```python
import pandas as pd
import numpy as np
import os
from tqdm.auto import tqdm

os.makedirs("../data_new/labels/", exist_ok=True)

feature_names = ["Habitat_ArtificalAquatic","Habitat_ArtificialTerrestrial","Habitat_Desert","Habitat_Forest","Habitat_Grassland","Habitat_Savanna","Habitat_Shrubland","Habitat_Wetland","Temp_Annual_Mean","Temp_Annual_Seasonality","Pre_Annual_Mean","Pre_Annual_Seasonality","Pre_Annual_DriestQuarter"]
feature_dfs = []
species_label_names = []

dfs = pd.read_csv("../data_new/habitat_bioclim_species01_passage_cleaned.csv", chunksize=1000)
for idx, df in tqdm(enumerate(dfs)):
    feature_dfs.append(df[feature_names])
    if len(species_label_names) == 0:
        species_label_names = [x for x in df.columns if str(x).startswith("X")]
    for name in species_label_names:
        output_path = f"../data_new/labels/{name}.csv"
        df[[name]].to_csv(output_path, mode="a", header=not os.path.exists(output_path), index=False)
pd.concat(feature_dfs).to_csv("../data_new/combined.csv", index=False)
```
